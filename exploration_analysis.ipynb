```python
age=df['Age'].dropna()
ageval=np.average(age)
ageval
```

```python
df['Age']=df['Age'].fillna(ageval)
```

```python
numeric_df = df.select_dtypes(include=['float64', 'int64'])
```

```python
numeric_df.info()
```

```python
correlation_matrix = numeric_df.corr() 
print(correlation_matrix)
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.show()
```

corelation matrix shows the relation between 2 features
